```python


patterns1=[]

for i in range(len(patterns)):
    pattern = patterns[i]
    print(pattern)
    p=[]
    for j in range(len(pattern)):
        print(pattern[j])
        measure = pattern[j]
        if measure =='0':
            p.append('R')
        elif measure == '1':
            p.append('L')
        elif measure == '2':
            p.append('RL')
        elif measure == '3':
            p.append('LR')
        else: print('error')
    t = tuple(p)
    patterns1.append(t)
```

```python
len(patterns1)
three = [2,6,7,12,18,22,25,26,29,32,37,39,43,45,46,54]

patterns2=[]
for i in range(len(patterns1)):
    if i+1 in three:
        patterns2.append(['3/4',patterns1[i]])
    elif len(patterns1[i])==6:
        patterns2.append(['2/4',patterns1[i]])
    else:
        if len(patterns1[i])==4:
            patterns2.append(['4/4',patterns1[i]])
        else: print('error')
```

```python
data = patterns2

occ = {}

for ts, pattern in data:
    if ts not in occ:
        occ[ts] = {}

    if pattern not in occ[ts]:
        occ[ts][pattern] = 0

    occ[ts][pattern] += 1

print(occ)
```
